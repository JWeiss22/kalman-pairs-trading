```python
import numpy as np
import pandas as pd
import yfinance as yf

prices = yf.download(["MA", "V"], start="2010-01-01", end="2026-09-01", auto_adjust=True)["Close"]
prices.head()
```

```python
prices.plot(figsize=(12, 5), title="MA vs V adjusted close")
```

```python
np.log(prices).plot(figsize=(12, 5), title="MA vs V log prices")
```
